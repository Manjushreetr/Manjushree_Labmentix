#### Chart - 2 - Stock Closing Prices by Year with Best Performing Year Highlighted

```python
df_yearly = df.groupby('Year').agg({'Difference': 'sum', 'Close': 'mean'}).reset_index()
```

The "Difference" column represents is the sum of gap between today's closing price and yesterday's closing price.

```python
df_yearly
```

```python
best_performing_year = df_yearly.loc[df_yearly['Difference'].idxmax()]
```

```python
best_performing_year
```

```python
plt.figure(figsize=(10, 5))
plt.plot(df_yearly['Year'], df_yearly['Close'], marker='o', linestyle='-', color='b', label='Average Close Price')


plt.scatter(best_performing_year['Year'], best_performing_year['Close'], color='red', label='Best Performing Year')
plt.xlabel('Year')
plt.ylabel('Average Close Price')
plt.title('Stock Closing Prices by Year with Best Performing Year Highlighted')
plt.legend()


plt.grid()
plt.show()
```

##### 1. Why did you pick the specific chart?

The line chart shows the trends at which the price has inclined or declined.

##### 2. What is/are the insight(s) found from the chart?

This chart indicates that 2017 was the best performing year.

##### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

We understand that the performance of Yes bank has declined after 2017.

#### Chart - 4 - Monthly Closing Price Differences for Each Year

```python
for year, data in df.groupby('Year'):
    plt.plot(data['Date'], data['Difference'], marker='o', linestyle='-', label=f'Year {year}')

plt.xlabel('Date')
plt.ylabel('Monthly Closing Price Difference')
plt.title('Monthly Closing Price Differences for Each Year')
plt.grid()
plt.tight_layout()
plt.show()
```

##### 1. Why did you pick the specific chart?

The line chart shows the trends at which the close price by each year.

##### 2. What is/are the insight(s) found from the chart?

This chart indicates that 2018 was the worst performing year.

##### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

From the chart above, we can observe that there was a significant decline in the closing price in 2018, after which the closing price continued to decrease.

#### Chart - 6 - Top 10 Best Performing Months

```python
df['Month'] = df['Datetime'].dt.month
```

```python
df['Month'].tail()
```

```python
monthly_performance = df.groupby('Month').agg({'Close': 'sum'}).reset_index()
```

```python
monthly_performance
```

```python
top_10_months = monthly_performance.sort_values(by='Close', ascending=False).head(10)
```

```python
plt.figure(figsize=(10, 6))
plt.bar(top_10_months['Month'].astype(str), top_10_months['Close'], color='skyblue')

plt.xlabel('Months')
plt.ylabel('Total Closing Price')
plt.title('Top 10 Best Performing Months Across All Years')

plt.tight_layout()
plt.show()
```

##### 1. Why did you pick the specific chart?

The Bar chart shows the trends at which the Closing Price with Highest Performing Months.

##### 2. What is/are the insight(s) found from the chart?

This chart indicates that July is the best performing average month in whole dataset.

##### 3. Will the gained insights help creating a positive business impact?

We can understand that the month July has done high performaance compared to other months.
